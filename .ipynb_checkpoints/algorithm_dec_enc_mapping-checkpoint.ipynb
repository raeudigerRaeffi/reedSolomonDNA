```python
from itertools import chain, combinations
import string
import time
import pickle
```

```python
def all_subsets(liste, s, t):
    return chain(*map(lambda x: combinations(liste, x), range(s, t)))
```

```python
doubles = []
for x in range(32,256):
    doubles.append((x,x))
    
numb_comb = list(all_subsets(list(range(32,256)), 2, 3))
reve_comb = []
for x in numb_comb:
    x = list(x)
    reve_comb.append((x[1],x[0]))

dec_comb = doubles+numb_comb+reve_comb
dec_comb.sort()
```

```python
numb_comb = list(all_subsets(list(range(0,47)), 3, 4))
reve1_comb, reve2_comb, reve3_comb, reve4_comb, reve5_comb = [],[],[],[],[]
for x in numb_comb:
    x = list(x)
    reve1_comb.append((x[0],x[2],x[1]))
    reve2_comb.append((x[1],x[0],x[2]))
    reve3_comb.append((x[1],x[2],x[0]))
    reve4_comb.append((x[2],x[0],x[1]))
    reve5_comb.append((x[2],x[1],x[0]))
    
tri_comb = numb_comb+reve1_comb+reve2_comb+reve3_comb
tri_comb.sort()
for x in range(len(tri_comb)-len(dec_comb)):
    tri_comb.pop()
```

```python
if len(tri_comb) == len(dec_comb):
    print('First mapping done')
```

```python
myDict = {k:v for k,v in zip(dec_comb, tri_comb)}
```
